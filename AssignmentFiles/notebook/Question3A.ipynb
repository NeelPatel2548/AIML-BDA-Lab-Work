```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.linear_model import LinearRegression, Ridge, Lasso, LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier

from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.decomposition import PCA

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, mean_absolute_error, mean_squared_error,
    r2_score, silhouette_score, adjusted_rand_score,
    pairwise_distances
)

from sklearn.metrics.pairwise import cosine_similarity
from scipy.cluster.hierarchy import linkage, dendrogram, cophenet
from scipy.spatial.distance import pdist


# ============================================================
# 1. LOAD AND PREPARE DATASET
# ============================================================

df = pd.read_csv("../data\Wine.csv")

X = df.drop("Customer_Segment", axis=1)
y = df["Customer_Segment"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_scaled = StandardScaler().fit_transform(X)


# Store final results
results = []

# 2. HELPER FUNCTION FOR CLASSIFICATION

def evaluate_model(name, model, scaled=False):

    if scaled:
        model.fit(X_train_scaled, y_train)
        predictions = model.predict(X_test_scaled)
    else:
        model.fit(X_train, y_train)
        predictions = model.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(
        y_test, predictions, average="weighted", zero_division=0
    )
    recall = recall_score(
        y_test, predictions, average="weighted", zero_division=0
    )
    f1 = f1_score(
        y_test, predictions, average="weighted", zero_division=0
    )

    results.append({
        "Task": "Classification",
        "Model": name,
        "R2": np.nan,
        "MAE": np.nan,
        "RMSE": np.nan,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Silhouette": np.nan,
        "ARI": np.nan
    })

    print(f"\n{name} Confusion Matrix:")
    print(confusion_matrix(y_test, predictions))

    return model

# 3. REGRESSION – LINEAR, RIDGE AND LASSO

X_reg = df.drop(columns=["Alcohol", "Customer_Segment"])
y_reg = df["Alcohol"]

Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_reg, y_reg,
    test_size=0.20,
    random_state=42
)

regression_models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": make_pipeline(
        StandardScaler(), Ridge(alpha=1.0)
    ),
    "Lasso Regression": make_pipeline(
        StandardScaler(), Lasso(alpha=0.1, max_iter=10000)
    )
}

for name, model in regression_models.items():

    model.fit(Xr_train, yr_train)
    pred = model.predict(Xr_test)

    mse = mean_squared_error(yr_test, pred)

    results.append({
        "Task": "Regression",
        "Model": name,
        "R2": r2_score(yr_test, pred),
        "MAE": mean_absolute_error(yr_test, pred),
        "RMSE": np.sqrt(mse),
        "Accuracy": np.nan,
        "Precision": np.nan,
        "Recall": np.nan,
        "F1": np.nan,
        "Silhouette": np.nan,
        "ARI": np.nan
    })

print("\n1(a) REGRESSION RESULTS")
display(pd.DataFrame([
    {
        "Model": r["Model"],
        "R2": r["R2"],
        "MAE": r["MAE"],
        "RMSE": r["RMSE"]
    }
    for r in results if r["Task"] == "Regression"
]).round(4))


# 4. DECISION TREE – ENTROPY AND GINI

tree_models = {
    "Decision Tree (Entropy)": DecisionTreeClassifier(
        criterion="entropy", random_state=42
    ),
    "Decision Tree (Gini)": DecisionTreeClassifier(
        criterion="gini", random_state=42
    )
}

entropy_tree = None

for name, model in tree_models.items():

    model = evaluate_model(name, model)

    if "Entropy" in name:
        entropy_tree = model

    tree = model.tree_

    feature_index = tree.feature[0]
    threshold = tree.threshold[0]

    left = 1
    right = 2
    total = tree.n_node_samples[0]

    child_impurity = (
        tree.n_node_samples[left] * tree.impurity[left]
        + tree.n_node_samples[right] * tree.impurity[right]
    ) / total

    gain = tree.impurity[0] - child_impurity

    print(f"\n{name}")
    print("Best split feature:", X.columns[feature_index])
    print("Threshold:", round(threshold, 4))

    if "Entropy" in name:
        print("Information Gain:", round(gain, 4))
    else:
        print("Gini Impurity Reduction:", round(gain, 4))


# Decision Tree diagram – only one
plt.figure(figsize=(18, 8))

plot_tree(
    entropy_tree,
    feature_names=X.columns,
    class_names=[str(c) for c in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title("Decision Tree using Entropy – Wine Dataset")
plt.tight_layout()
plt.show()

# 5. LOGISTIC REGRESSION – L2 AND L1

logistic_models = {
    "Logistic Regression (Ridge/L2)": LogisticRegression(
        penalty="l2", max_iter=10000, random_state=42
    ),
    "Logistic Regression (Lasso/L1)": LogisticRegression(
        penalty="l1", solver="saga",
        max_iter=20000, random_state=42
    )
}

for name, model in logistic_models.items():

    model = make_pipeline(StandardScaler(), model)
    model = evaluate_model(name, model, scaled=False)


# Predicted probabilities from L2 model
l2_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(penalty="l2", max_iter=10000)
)

l2_model.fit(X_train, y_train)

print("\nLOGISTIC REGRESSION – FIRST FIVE PREDICTED PROBABILITIES")

display(pd.DataFrame(
    l2_model.predict_proba(X_test.iloc[:5]),
    columns=[f"Class {c}" for c in l2_model.classes_]
).round(4))


# Logistic decision boundary using two PCA components
pca_2d = PCA(n_components=2)
X_2d = pca_2d.fit_transform(X_scaled)

Xb_train, Xb_test, yb_train, yb_test = train_test_split(
    X_2d, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

boundary_model = LogisticRegression(max_iter=10000)
boundary_model.fit(Xb_train, yb_train)

x_min, x_max = X_2d[:, 0].min() - 1, X_2d[:, 0].max() + 1
y_min, y_max = X_2d[:, 1].min() - 1, X_2d[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 250),
    np.linspace(y_min, y_max, 250)
)

Z = boundary_model.predict(
    np.c_[xx.ravel(), yy.ravel()]
).reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, alpha=0.25)

for c in sorted(y.unique()):
    mask = y.values == c
    plt.scatter(
        X_2d[mask, 0], X_2d[mask, 1],
        label=f"Class {c}"
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Logistic Regression Decision Boundary")
plt.legend()
plt.tight_layout()
plt.show()

# 6. K-NEAREST NEIGHBORS

knn = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5)
)

knn = evaluate_model("KNN (k=5)", knn)


# 7. GAUSSIAN NAIVE BAYES

nb = GaussianNB()
nb = evaluate_model("Gaussian Naive Bayes", nb)

# 8. DISTANCE AND SIMILARITY
print("\nEUCLIDEAN DISTANCE MATRIX – FIRST FIVE RECORDS")
display(pd.DataFrame(
    pairwise_distances(X_scaled[:5], metric="euclidean")
).round(3))

print("\nMANHATTAN DISTANCE MATRIX – FIRST FIVE RECORDS")
display(pd.DataFrame(
    pairwise_distances(X_scaled[:5], metric="manhattan")
).round(3))

print("\nCOSINE SIMILARITY MATRIX – FIRST FIVE RECORDS")
display(pd.DataFrame(
    cosine_similarity(X_scaled[:5])
).round(3))


# 9. K-MEANS CLUSTERING
kmeans_scores = []

for k in range(2, 8):

    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = km.fit_predict(X_scaled)

    kmeans_scores.append({
        "K": k,
        "Silhouette": silhouette_score(X_scaled, labels),
        "Inertia": km.inertia_
    })

kmeans_table = pd.DataFrame(kmeans_scores)

best_k = int(
    kmeans_table.loc[kmeans_table["Silhouette"].idxmax(), "K"]
)

kmeans = KMeans(
    n_clusters=best_k,
    n_init=10,
    random_state=42
)

kmeans_labels = kmeans.fit_predict(X_scaled)

km_sil = silhouette_score(X_scaled, kmeans_labels)
km_ari = adjusted_rand_score(y, kmeans_labels)

print("\nK-MEANS RESULTS")
print("Best K:", best_k)
print("Silhouette:", round(km_sil, 4))
print("ARI:", round(km_ari, 4))

results.append({
    "Task": "Clustering",
    "Model": "K-Means",
    "R2": np.nan, "MAE": np.nan, "RMSE": np.nan,
    "Accuracy": np.nan, "Precision": np.nan,
    "Recall": np.nan, "F1": np.nan,
    "Silhouette": km_sil, "ARI": km_ari
})



# 10. HIERARCHICAL CLUSTERING

methods = ["single", "complete", "average", "ward"]

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
cophenetic_results = []

for method, ax in zip(methods, axes.ravel()):

    Z_link = linkage(X_scaled, method=method)

    corr, _ = cophenet(Z_link, pdist(X_scaled))

    cophenetic_results.append({
        "Linkage": method,
        "Cophenetic Correlation": corr
    })

    dendrogram(
        Z_link,
        truncate_mode="lastp",
        p=15,
        ax=ax
    )

    ax.set_title(f"{method.title()} Linkage")
    ax.set_xlabel("Cluster or Sample")
    ax.set_ylabel("Distance")

plt.tight_layout()
plt.show()

print("\nHIERARCHICAL LINKAGE RESULTS")
display(pd.DataFrame(cophenetic_results).round(4))

hierarchical = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

hier_labels = hierarchical.fit_predict(X_scaled)

hier_sil = silhouette_score(X_scaled, hier_labels)
hier_ari = adjusted_rand_score(y, hier_labels)

print("Hierarchical Silhouette:", round(hier_sil, 4))
print("Hierarchical ARI:", round(hier_ari, 4))

results.append({
    "Task": "Clustering",
    "Model": "Hierarchical Clustering",
    "R2": np.nan, "MAE": np.nan, "RMSE": np.nan,
    "Accuracy": np.nan, "Precision": np.nan,
    "Recall": np.nan, "F1": np.nan,
    "Silhouette": hier_sil, "ARI": hier_ari
})

# 11. RANDOM FOREST AND BOOSTING
try:
    ada = AdaBoostClassifier(
        estimator=DecisionTreeClassifier(max_depth=1),
        n_estimators=100,
        random_state=42
    )
except TypeError:
    ada = AdaBoostClassifier(
        base_estimator=DecisionTreeClassifier(max_depth=1),
        n_estimators=100,
        random_state=42
    )

ensemble_models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=200, random_state=42
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100, random_state=42
    ),
    "AdaBoost": ada
}

for name, model in ensemble_models.items():
    evaluate_model(name, model)

# 12. GAUSSIAN MIXTURE MODEL
gmm_scores = []

for n in range(2, 8):

    model = GaussianMixture(
        n_components=n,
        random_state=42,
        n_init=5
    )

    model.fit(X_scaled)

    gmm_scores.append({
        "Components": n,
        "AIC": model.aic(X_scaled),
        "BIC": model.bic(X_scaled)
    })

gmm_table = pd.DataFrame(gmm_scores)

best_n = int(
    gmm_table.loc[gmm_table["BIC"].idxmin(), "Components"]
)

gmm = GaussianMixture(
    n_components=best_n,
    random_state=42,
    n_init=5
)

gmm_labels = gmm.fit_predict(X_scaled)

gmm_sil = silhouette_score(X_scaled, gmm_labels)
gmm_ari = adjusted_rand_score(y, gmm_labels)

print("\nGAUSSIAN MIXTURE MODEL RESULTS")
print("Best Components:", best_n)
print("Silhouette:", round(gmm_sil, 4))
print("ARI:", round(gmm_ari, 4))

results.append({
    "Task": "Clustering",
    "Model": "Gaussian Mixture Model",
    "R2": np.nan, "MAE": np.nan, "RMSE": np.nan,
    "Accuracy": np.nan, "Precision": np.nan,
    "Recall": np.nan, "F1": np.nan,
    "Silhouette": gmm_sil, "ARI": gmm_ari
})


# 13. PRINCIPAL COMPONENT ANALYSIS

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

explained = pca.explained_variance_ratio_

print("\nPCA RESULTS")
print("Explained Variance Ratio:", np.round(explained, 4))
print("Total Variance Explained:", round(explained.sum(), 4))

plt.figure(figsize=(8, 6))

for c in sorted(y.unique()):
    mask = y.values == c
    plt.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        label=f"Class {c}"
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Wine Dataset")
plt.legend()
plt.tight_layout()
plt.show()

# 14. FINAL MODEL PERFORMANCE COMPARISON

final_table = pd.DataFrame(results)

print("\nFINAL MODEL PERFORMANCE COMPARISON")

display(final_table.round(4).reset_index(drop=True))

# Best classification model
classification_table = final_table[
    final_table["Task"] == "Classification"
]

best = classification_table.loc[
    classification_table["Accuracy"].idxmax()
]

print("\nBEST CLASSIFICATION MODEL:", best["Model"])
print("Accuracy:", round(best["Accuracy"], 4))
print("Precision:", round(best["Precision"], 4))
print("Recall:", round(best["Recall"], 4))
print("F1 Score:", round(best["F1"], 4))
```

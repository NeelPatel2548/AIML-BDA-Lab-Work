```python
#class demo
class Person:
    ins = "MSIS"

    def name(self, newname):
        self.name = newname;
        print(self.name)
    def loc(self, newloc):
        self.loc = newloc;
        print(self.loc)


per = Person()
per.name("User")
per.loc("Manipal")

#remeber the structur
print(
    '{} is a student of {} institue and living in {} city' .format(per.name, per.ins, per.loc)
)
```

```python
def calSqu(num):
    return num * num

num = (1,2,3,4,5,6,10)
res = map(calSqu, num)
print(res)

result = list(res)
print(result)

# #same result
# for i in result:
#     print(i)
```

```python
num = (1,2,3,4,5,6)

#### LAMBDA FUNCTION ONLY WORK FOR INLINE FUNCTION NOT OUTSIDE OF SINGLE LINE CODE

res = map(lambda x: x*x, num)
print(list(res))
```

```python
ppl = ['ARG1', 'ARG2']

def spliting(ppl):
    arg = ppl.split()[0]
    arg1 = ppl.split()[-1]

    return '{} {}'.format(arg, arg1)

list(spliting(ppl))
```

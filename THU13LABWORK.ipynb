```python
# 3. Create a NumPy array containing the numbers from 10 to 50. Display: 

arr = np.arange(10,51)

 
# The array 
print(arr)

# Number of elements 
print("No of ele: ", len(arr))

# Shape 
n = arr.shape
print("Shape:", n)

# Dimension 
arr.resize((11, 4))
print("Resizing : ",arr)

# Data type 
print("Datatype: ",type(arr))
```

```python
# 4. Using NumPy, create: 
 
# A 1D array containing 10 zeros 
dd =  np.zeros([10],int) 
print("10 Zeors: ", dd)

# A 3 × 3 array containing ones 
three =  np.ones([3,3], int) 
print("3 * 3: \n", three)

# A 4 × 4 identity matrix 
iden =  np.eye(4)
print("Identical matrix: \n", iden)

# An array containing numbers from 5 to 50 with a step size of 5. 
ar = np.arange(5,51,5)
print("Skiping: ", ar)
```

```python
# 5. Create an array containing numbers from 1 to 24. Reshape it into: 

ar = np.arange(1,25)

# 4 × 6 
print("4 * 6: ")
ar.resize((4,6))
print(ar)

# 6 × 4 
print("\n6 * 4: ")
ar.resize((6,4))
print(ar)

# 2 × 3 × 4 
print("\n2 * 3 * 4: ")
ar.resize((2,3,4))
print(ar)


# Display the shape and ndim of each result.
ar.shape
```

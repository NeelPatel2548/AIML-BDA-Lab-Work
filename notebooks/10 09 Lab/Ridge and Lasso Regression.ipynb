```python
# 1. before converting data into ridge regression, we have to do StandardScalar first

# #Create pipeline:

# 1. Standardize the predicitons
# 2. 





import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
data = {
    "Study_hour": [2,3,4,5,6,7,8,9],
    "aten" : [60,65,70,75,80,85,90,95],
    "ass_sc" : [50,55,58,65,70,74,80,88],
    "final_mark" : [42,48,54,61,67,73,81,89]
}

df = pd.DataFrame(data)

x = df.iloc[:,0:3]
y = df["final_mark"]
```

```python
x
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
scaler = StandardScaler()
X_scaled = scaler.fit_transform(x)             # ---- X is our independen varibale
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = make_pipeline(
    StandardScaler(),
    Ridge(alpha=1.0)
)

model.fit(x_train, y_train)
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.linear_model import RidgeCV

model = RidgeCV(
    alphas= [0.01, 0.1, 1, 10,100],
    cv = 5
)

model.fit(x_train, y_train)

print("Best Alphas: ", model.alpha_)
```

```python
model = make_pipeline(
    StandardScaler(),
    RidgeCV(alphas= [0.01, 0.1, 1, 10,100])
)

model.fit(x_train, y_train)

Ridge_model = model.named_steps["ridgecv"]

print("Best Alphas: ", Ridge_model.alpha_)
```

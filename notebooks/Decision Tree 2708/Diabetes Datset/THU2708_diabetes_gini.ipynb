```python
# for below cells parameters are:
#     1. gini
#     2. 0.2

# # dataset = "Iris Dataset"

import pandas as ps
import numpy as nm
import matplotlib.pyplot as plt

df = ps.read_csv("D:\BDA 20\AIML-BDA-Lab-Work\data\diabetes.csv")
df.head()


# df.info()
# print(df.isnull().sum())

# print("Independent variable, X:")
x = df.iloc[:, 0:8].values  #Pregnancies,Glucose,BloodPressure,SkinThickness,Insulin,BMI,DiabetesPedigreeFunction,Age
# print(x)

# print("Dependent variable, Y:")
y = df.iloc[:, 8].values      #Outcome
# print(y)

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y,
    test_size=0.2,
    random_state=45
)

from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train) 
x_test=sc_x.transform(x_test)

# print(x_test)

from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion='gini', random_state=0)

classifier.fit(x_train,y_train)


y_pred = classifier.predict(x_test)        

# print("Actual: ",y_test)
# print("Predicted: ",y_pred)

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)

print("\n\nTrue Pos", cm[0][0])
print("True Neg", cm[0][1])
print("False Pos", cm[1][0])
print("False Neg", cm[1][1])

accuracy = (cm[0][0]  + cm[1][1]) / (cm[0][0]  + cm[1][1] + cm[1][0] + cm[0][1])
print("\nAccuracy - ", accuracy * 100, "%")

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))

from sklearn import tree
tree.plot_tree(classifier)
```

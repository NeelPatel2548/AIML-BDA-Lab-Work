```python
import pandas as pd
# Pandas Assignment

# 1. Create a Pandas Series containing the marks: 78, 85, 92, 67, 88

#     Display:

# 	Complete Series
liist = [45,78,98,56,65,76,48]

series = pd.Series(liist)

print("Display series: \n",series)

# 	Values
print("Values: ", series.values)

# 	Index
print("Index: ", series.index)

# 	Data type
print("DataType: ", type(series))

# 	Mean
print("Mean: ", series.mean())

# 	Maximum
print("Max: ", series.max())

# 	Minimum
print("Min: ", series.min())


```

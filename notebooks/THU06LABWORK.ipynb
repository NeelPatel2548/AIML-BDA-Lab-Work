```python
##Task 2 - Write code using lambda function that takes in three parameters and adds the first two.

res = lambda a,b,c : (a+b)
cal = res(5,10,5)

print(cal)
```

```python
##Task 3 - Write your own program to show the use of lambda functions

res = lambda num: "Positive" if num > 0 else "Negative"

print(res(-45))
print(res(10))
```

```python
## Task 4 - Program to double each item in a list using map()

num = [1,2,3,4,5,2,4]

res = map(lambda num: num*2, num)

print(list(res))
```

```python
##Task 6 - computes square numbers using mapping function/ lambda function
num = [1,2,3,4,5,2,4]

res = map(lambda num: num*num, num)

print(list(res))
```

```python
##Task 7 - Compute square of numbers using list comprehension technique.

num = [1,2,3,4,5,2,4]

res = [i * i for i in num]

print(list(res))
```

```python
##Task 8 - Compute square of even numbers using  using list comprehension technique.

num = [1,2,3,4,5,2,4]

res = [i * i for i in num if i % 2 == 0]

print(list(res))
```

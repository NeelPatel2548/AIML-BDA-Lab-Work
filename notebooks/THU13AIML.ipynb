```python
#use of Vstack and stack arrays in squencse vertically (row wise)
# p = [[1,1,1],
#      [1,1,1],
#      [1,1,1]]

p = np.ones([3,3], int)         #used to make 1's array like i create above

np.vstack([p,2*p]) #vertical stack

np.hstack([p, 2*p, 5*p, 6*p]) #horizontal stack
```

```python
x = np.array([10,42,34])
y = np.array([41,25,16])

#basic arithmetic operation on array list

print(x+y)
print(x-y)

print(x*y)
print(x/y)

print(x % y)

#Dot product :
print(x@y)   # @ use for dot product like: 1*4 + 2*5 + 3*6 = 32
print(x.dot(y))

#power operation:
z= np.array([y,y**2])
print(z)

#to transport the matrix
print(z.T)

print(z.T.shape)


#use .astype to cast to a specific type
z.astype('f')
```

```python
#math built in functions

math = np.array([-5,6,8,9,10])
            #     0,1,2,3,4
            #     -5,-4,-3,-2,-1
print(math.sum())

print(math.min())
print(math.max())

#return index of max and min element
print(math.argmax())
print(math.argmin())
```

```python
#indexing and slicing

arr = np.arange(10)     #its going to print 0-10 numbers
print(arr)

# np.arange(10)**2        it takes 0 to 10 and multiply with 2
# arr[-1]                     it print last index element

print(arr[1:8])             #start at 1 and end at secon last like: 8-1 = 7
print(arr[2::2])            #In a NumPy array, the :: syntax is the extended slicing operator used to skip elements by a specified step size.
```

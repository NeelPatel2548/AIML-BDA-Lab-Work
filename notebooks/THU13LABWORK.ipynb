```python
# 8. Given:
salary = np.array([25000, 32000, 28000, 45000,50000, 38000, 42000, 30000])

# Find:
# Total salary
print("Sum of salary: ",salary.sum())

# Average salary
print("Average Salary: ",np.average(salary))

# Highest salary
print("Highest Salary: ",salary.max())

# Lowest salary
print("Lowest salary: ", salary.min())

# Median salary
print("Median Salary: ", np.median(salary))

# Standard deviation
print(f"standard Deviation: {np.std(salary):.2f}")


```

```python
# 10. The following array contains marks of four students in three subjects:

marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

# Total marks of each student
res = marks.sum(axis=1)
print("Total marks of each student: ",res)

# Average marks of each student
res1 = marks.mean(axis=1)
print("Average marks of each student: ",res1)

# Average marks in each subject
res2 = marks.mean(axis=0)
print("Average marks in each subject: ",res2)

# Student with the highest total
res3 = np.argmax(res)
print("Student with the highest total: ",res3)

# Highest mark in each subject
res4 = marks.max(axis=0)
print("Highest mark in each subject: ",res4)
```

```python
# 12. Daily temperatures for two weeks are:

temperature = np.array([
31, 32, 33, 35, 36, 34, 32,
30, 29, 31, 33, 35, 37, 36
])

# Find:
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

# Mean temperature
res = np.mean(temperature)
print("Mean temperature: ",res)

# Maximum and minimum temperatures
res1 = np.max(temperature)
res2 = np.min(temperature)
print("Maximum temperature: ",res1)
print("Minimum temperature: ",res2)

# Number of days above the average temperature
res3 = np.sum(temperature > res)
print("Days above average: ",res3)

# Temperatures greater than 34°C
res4 = temperature[temperature > 34]
print("Temperatures > 34: ",res4)

# Difference between maximum and minimum
res5 = res1 - res2
print("Max - Min difference: ",res5)

# Standard deviation
res6 = np.std(temperature)
print("Standard deviation: ",res6)
```

```python
# 13. Monthly sales of a company are:
sales = np.array([
12000, 15000, 17000, 14000,
18000, 21000, 19000, 22000,
25000, 23000, 26000, 28000
])
# Determine:
# Total annual sales
res = np.sum(sales)
print("Total annual sales: ",res)

# Average monthly sales
res1 = np.mean(sales)
print("Average monthly sales: ",res1)

# Month with the highest sales
res2 = np.argmax(sales) + 1
print("Month with highest sales: ",res2)

# Month with the lowest sales
res3 = np.argmin(sales) + 1
print("Month with lowest sales: ",res3)

# Number of months with sales above ₹20,000
res4 = np.sum(sales > 20000)
print("Months with sales above 20000: ",res4)

# Percentage increase from January to December
res5 = ((sales[-1] - sales[0]) / sales[0]) * 100
print("Percentage increase Jan to Dec: ",res5, "%")
```

```python
# we have 2 kind of variable: 1. Indepenent var and 2. Dependent var.
# we do: Column split for it then ROW split.


#split independent variable
#loc uses labels
# iloc uses integer position (number from 0 to the length of the data) 

print("InDependent variable, X:")
x = df.iloc[:, :-1].values                  #[:, :-1] --- all column; not last one
print(x)

print("Dependent variable, Y:")
y = df.iloc[:, 3 ].values
print(y)
```

```python
#taking care of missing data

from sklearn.impute import SimpleImputer

impute = SimpleImputer(missing_values=nm.nan, strategy='mean')
impute = impute.fit(x[:, 1:3])                      #fit: is going to perform operation on dataset of terms we give to it.

x[:, 1:3] = impute.transform(x[:, 1:3])
print(x)
```

```python
#Encoding categorial data
#Encoding the Indpendent variable
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
labelencoder_x = LabelEncoder()
x[:, 0] = labelencoder_x.fit_transform(x[:,0])

print(x)
```

```python
#Split data into Training and Test data

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=45)      #test_size = 0.2 --- it take 80% data for trainig and other 20% for testing

print("Train: ", x_train, "\n\n\nTest:", x_test,"\n\n\nY Train: ", y_train, "\n\n\nY Test:", y_test)
```

```python
#Feature scaling 
#sc_X = (X[1]-mean) / Standard deavition
#For classification we are going to use Scaling or StandardScaler.
#For regression we are not going to use StandardScaler.

from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train)               #fit_transform --- its going to 'learn'
x_test = sc_x.transform(x_test)


print("Train: ",x_train,"\n\n\nTest: ",x_test)
```

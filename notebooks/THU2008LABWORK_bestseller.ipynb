```python
import pandas as ps
import numpy as nm
import matplotlib.pyplot as plt
```

```python
#importing data:

df = ps.read_csv("../data/bestsellers.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()

# null_rows = df[df.isnull().all(axis=1)]
# print(null_rows)
```

```python
print("Independent variable, X:")
x = df.iloc[:, [2, 3, 4, 5]].values     # User Rating, Reviews, Price, Year
print(x)

print("Dependent variable, Y:")
y = df.iloc[:, 6].values                 # Genre (Fiction / Non Fiction)
print(y)
```

```python
print("Duplicate Values: ", df.duplicated().sum())
```

```python
from sklearn.impute import SimpleImputer
import numpy as nm

#handling missing value of X
impute = SimpleImputer(missing_values=nm.nan, strategy='mean')
x = impute.fit_transform(x)

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=45)

```

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train)
x_test = sc_x.transform(x_test)

from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state=45)
classifier.fit(x_train, y_train)

y_pred = classifier.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy * 100, "%")
print(classification_report(y_test, y_pred))
```
